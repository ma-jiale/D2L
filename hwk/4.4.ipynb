```python
def cross_entropy_from_concept(y_hat, y):
    cross_entropy = 0
    for i in range(len(y_hat)):
        for j in range(len(y_hat[0])):
            cross_entropy += -(y[i][j] * torch.log(y_hat[i][j]))
    return cross_entropy / (len(y_hat))
```
