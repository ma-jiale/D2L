1. Give an example where observing more data can reduce the amount of uncertainty about the outcome to an arbitrarily low level.
   请举出一个例子，说明观测更多数据可以将结果的不确定性降低到任意低的水平。

2. Give an example where observing more data will only reduce the amount of uncertainty up to a point and then no further. Explain why this is the case and where you expect this point to occur.
   请举出一个例子，说明观测更多数据只能将不确定性降低到某个程度，之后便不再降低。请解释为什么会这样，以及你预期这个极限点会出现在哪里。

3. We empirically demonstrated convergence to the mean for the toss of a coin. Calculate the variance of the estimate of the probability that we see a head after drawing $n$ samples.
   我们通过实验验证了抛硬币时均值收敛的现象。请计算在抽取 $n$ 个样本后，对出现正面概率的估计值的方差。

   1. How does the variance scale with the number of observations?
      方差如何随观测数量变化？

   2. Use Chebyshev's inequality to bound the deviation from the expectation.
      使用切比雪夫不等式（Chebyshev's inequality）对偏离期望值的情况进行上界估计。

   3. How does it relate to the central limit theorem?
      这与中心极限定理（Central Limit Theorem）有什么关系？

4. Assume that we draw $m$ samples $x_i$ from a probability distribution with zero mean and unit variance. Compute the averages $z_m \stackrel{\text{def}}{=} m^{-1} \sum_{i=1}^m x_i$. Can we apply Chebyshev's inequality for every $z_m$ independently? Why not?
   假设我们从一个均值为零、方差为1的概率分布中抽取 $m$ 个样本 $x_i$。计算平均值 $z_m \stackrel{\text{def}}{=} m^{-1} \sum_{i=1}^m x_i$。我们能否对每个 $z_m$ 独立地应用切比雪夫不等式？为什么不能？

5. Given two events with probability $P(\mathcal{A})$ and $P(\mathcal{B})$, compute upper and lower bounds on $P(\mathcal{A} \cup \mathcal{B})$ and $P(\mathcal{A} \cap \mathcal{B})$. Hint: graph the situation using a [Venn diagram](https://en.wikipedia.org/wiki/Venn_diagram).
   给定两个事件，其概率分别为 $P(\mathcal{A})$ 和 $P(\mathcal{B})$，请计算 $P(\mathcal{A} \cup \mathcal{B})$ 和 $P(\mathcal{A} \cap \mathcal{B})$ 的上界和下界。提示：使用[维恩图](https://en.wikipedia.org/wiki/Venn_diagram)来分析。

6. Assume that we have a sequence of random variables, say $A$, $B$, and $C$, where $B$ only depends on $A$, and $C$ only depends on $B$, can you simplify the joint probability $P(A, B, C)$? Hint: this is a [Markov chain](https://en.wikipedia.org/wiki/Markov_chain).
   假设我们有一个随机变量序列，例如 $A$、$B$ 和 $C$，其中 $B$ 仅依赖于 $A$，$C$ 仅依赖于 $B$，你能否简化联合概率 $P(A, B, C)$？提示：这是一个[马尔可夫链](https://en.wikipedia.org/wiki/Markov_chain)。

7. In :numref:`subsec_probability_hiv_app`, assume that the outcomes of the two tests are not independent. In particular assume that either test on its own has a false positive rate of 10% and a false negative rate of 1%. That is, assume that $P(D=1 \mid H=0) = 0.1$ and that $P(D=0 \mid H=1) = 0.01$. Moreover, assume that for $H=1$ (infected) the test outcomes are conditionally independent, i.e., that $P(D_1, D_2 \mid H=1) = P(D_1 \mid H=1)\, P(D_2 \mid H=1)$ but that for healthy patients the outcomes are coupled via $P(D_1 = D_2 = 1 \mid H=0) = 0.02$.
   在 :numref:`subsec_probability_hiv_app` 中，假设两次检测的结果不是独立的。具体而言，假设每次检测单独的假阳性率为 10%，假阴性率为 1%。即假设 $P(D=1 \mid H=0) = 0.1$，且 $P(D=0 \mid H=1) = 0.01$。此外，假设当 $H=1$（已感染）时，两次检测结果条件独立，即 $P(D_1, D_2 \mid H=1) = P(D_1 \mid H=1)\, P(D_2 \mid H=1)$；但对于健康患者，两次检测结果通过 $P(D_1 = D_2 = 1 \mid H=0) = 0.02$ 耦合在一起。

   1. Work out the joint probability table for $D_1$ and $D_2$, given $H=0$ based on the information you have so far.
      根据已有信息，推导出在 $H=0$ 条件下 $D_1$ 和 $D_2$ 的联合概率表。

   2. Derive the probability that the patient is diseased ($H=1$) after one test returns positive. You can assume the same baseline probability $P(H=1) = 0.0015$ as before.
      推导在一次检测返回阳性后，患者确实患病的概率（$H=1$）。你可以假设与之前相同的基础概率 $P(H=1) = 0.0015$。

   3. Derive the probability that the patient is diseased ($H=1$) after both tests return positive.
      推导在两次检测都返回阳性后，患者确实患病的概率（$H=1$）。

8. Assume that you are an asset manager for an investment bank and you have a choice of stocks $s_i$ to invest in. Your portfolio needs to add up to $1$ with weights $\alpha_i$ for each stock. The stocks have an average return $\boldsymbol{\mu} = E_{\mathbf{s} \sim P}[\mathbf{s}]$ and covariance $\boldsymbol{\Sigma} = \text{Cov}_{\mathbf{s} \sim P}[\mathbf{s}]$.
   假设你是一家投资银行的资产管理人，你可以选择股票 $s_i$ 进行投资。你的投资组合权重 $\alpha_i$ 之和必须等于 1。这些股票的平均收益为 $\boldsymbol{\mu} = E_{\mathbf{s} \sim P}[\mathbf{s}]$，协方差矩阵为 $\boldsymbol{\Sigma} = \text{Cov}_{\mathbf{s} \sim P}[\mathbf{s}]$。

   1. Compute the expected return for a given portfolio $\boldsymbol{\alpha}$.
      计算给定投资组合 $\boldsymbol{\alpha}$ 的期望收益。

   2. If you wanted to maximize the return of the portfolio, how should you choose your investment?
      如果你想最大化投资组合的收益，应该如何选择投资？

   3. Compute the *variance* of the portfolio.
      计算投资组合的*方差*。

   4. Formulate an optimization problem of maximizing the return while keeping the variance constrained to an upper bound. This is the Nobel-Prize winning [Markowitz portfolio](https://en.wikipedia.org/wiki/Markowitz_model) :cite:`Mangram.2013`. To solve it you will need a quadratic programming solver, something way beyond the scope of this book.
      构建一个优化问题：在方差约束不超过某个上界的前提下最大化收益。这就是获得诺贝尔奖的 [Markowitz 投资组合模型](https://en.wikipedia.org/wiki/Markowitz_model) :cite:`Mangram.2013`。求解它需要一个二次规划求解器，这远远超出了本书的范围。
