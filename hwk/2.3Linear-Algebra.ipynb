```python
LA = torch.randn(2**10, 2**16)
LB = torch.randn(2**16, 2**5)
LC = torch.randn(2**5, 2**16)
LC = LB.T
```

```python
(LA@LB) @ LC
```

```python
LA@ (LB @ LC)
```

```python
import timeit
import tracemalloc
import gc

# 预热，避免首次运行的初始化开销影响结果
_ = (LA @ LB) @ LC

# --------------------
# 测量执行时间
# --------------------
repeat = 5
number = 10

times = timeit.repeat(
    stmt="(LA @ LB) @ LC",
    globals={"LA": LA, "LB": LB, "LC": LC},
    repeat=repeat,
    number=number
)

best_avg_time = min(times) / number
mean_avg_time = sum(times) / (repeat * number)

print(f"最佳单次耗时：{best_avg_time:.6f} 秒")
print(f"平均单次耗时：{mean_avg_time:.6f} 秒")

# --------------------
# 测量内存
# --------------------
gc.collect()
tracemalloc.start()

before_current, _ = tracemalloc.get_traced_memory()

result = (LA @ LB) @ LC

after_current, peak = tracemalloc.get_traced_memory()
tracemalloc.stop()

net_memory = after_current - before_current
peak_memory = peak - before_current

print(f"内存净增加：{net_memory / 1024**2:.3f} MiB")
print(f"执行期间峰值增加：{peak_memory / 1024**2:.3f} MiB")
```
